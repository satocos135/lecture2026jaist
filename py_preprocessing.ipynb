```python
# トークン用のクラスを定義

from typing import NamedTuple


class Token(NamedTuple):
    surface: str  # 表層形
    pos: str  # 品詞: POS part of speech
    subtype: str  # 品詞細分類
    lemma: str  # 原型
```

```python
def convert(row: str) -> Token:
    # 1行ずつ処理する
    surface, body = row.split('\t')
    splitted = body.split(',')
    pos = splitted[0]
    subtype = splitted[1]
    lemma = splitted[-3]
    return Token(surface, pos, subtype, lemma)
```

```python
[convert(x) for x in parsed.split('\n')[:-2]]
```

```python
def tokenize(text: str, tagger: MeCab.Tagger = tagger) -> list[Token]:
    # これも関数化しておく
    parsed = tagger.parse(text)
    return [convert(x) for x in parsed.split('\n')[:-2]]
```

```python
text = '切ってきて、丸めて、家の屋根にしげておくん。それはてんかごめんけどどこのやまいってきってきてもよいということになっちょった。'
tokenize(text)
```

```python
# 表現を変える
text = '切ってきて、丸めて、家の屋根にすげておくん。それは天下御免でどこの山に行って切ってきてもよいということになっていた。'
tokenize(text)
```

```python
# 辞書登録前
text = 'まあそういうことで、余分な話はいいけん。'
tokenize(text)
```
